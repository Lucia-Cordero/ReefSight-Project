# Load Pre-trained Model and Predict Using a Specific Image

```python
# Import necessary libraries
import io
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import tensorflow as tf
from tensorflow.keras.preprocessing import image
```

```python
# Define paths and variables for model and image classification
# Path to the pre-trained model
model_path = '/Users/et/code/Lucia-Cordero/ReefSight-Project/models/VGG16_image_model.keras'

# Path to the test image for prediction
img_path = '/Users/et/code/Lucia-Cordero/ReefSight-Project/raw_data/image_data/raw_image_data/train_validation_test/test/Bleached/3.png'

# List of class names for the model output
class_names = ['Bleached', 'Unbleached']
```

```python
# Load the pre-trained model
model_loaded = tf.keras.models.load_model(model_path)

# Load and preprocess the image
with open(img_path, 'rb') as file:
    img_data = file.read()

img = Image.open(io.BytesIO(img_data))
img = img.convert('RGB')  # Convert image to RGB format
img = img.resize((224, 224))  # Resize image to match model input dimensions

# Convert image to array and reshape for prediction
img_array = image.img_to_array(img)
img_array = img_array.reshape((-1, 224, 224, 3))

# Make predictions using the loaded model
predictions = model_loaded.predict(img_array)
print(f"Probability Unbleached: {predictions[0][0]}")
print(f"Probability Bleached: {1 - predictions[0][0]}")

# Determine the predicted class
predicted_class = (predictions[0][0] >= 0.5).astype("int32")
predicted_label = class_names[predicted_class]
print(f"Predicted class index: {predicted_class}\nPredicted class: {predicted_label}")

# Visualize the input image
plt.imshow(img)
plt.axis('off')  # Hide axes ticks
plt.show()
```
